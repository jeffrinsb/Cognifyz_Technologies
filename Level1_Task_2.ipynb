```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Data_set 2 - Copy.csv')
```

```python
 # To identify the gender column
df.columns
```

```python
gender_counts = df['gender'].value_counts()
```

```python
# Bar Chart
plt.figure(figsize=(6, 4))
gender_counts.plot(kind='bar', color=['skyblue', 'pink'])
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Gender Distribution - Bar Chart")
plt.xticks(rotation=0)
plt.show()
```
