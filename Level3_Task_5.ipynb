```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Data_set 2 - Copy.csv')
print(df)
```

```python
reason_columns = ['Reason_Equity', 'Reason_Mutual', 'Reason_Bonds', 'Reason_FD']
```

```python
all_reasons = pd.concat([df[col] for col in reason_columns], ignore_index=True).dropna()
reason_counts = all_reasons.value_counts()
```

```python
print("Top Reasons for Investment:\n", reason_counts.head(10))
```

```python
print("\nSummary: People commonly invest for", ', '.join(reason_counts.index[:5]))
```
