```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df['Duration'] = pd.to_numeric(df['Duration'], errors='coerce')
```

```python
avg_duration = df['Duration'].mean()
```

```python
print(f"Average Investment Duration: {avg_duration:.2f} years")
```
